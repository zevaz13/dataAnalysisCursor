# Explore channels

Plot individual or groups of channels in the time and frequency domains.

Available channels in this recording: `PO7 O1 Oz O2 PO8 PO4 POz PO3 P5 P3 P1 Pz P2 P4 P6 CP6 CP2 CPz CP1 CP5 C5 FC5 FC1 Cz FC2 F4 FC6 C6 Fz F3 AF3 AF4`

```python
%matplotlib inline
%load_ext autoreload
%autoreload 2

import sys
from pathlib import Path

PROJECT_ROOT = Path("../..").resolve()
SRC = PROJECT_ROOT / "src"
if str(SRC) not in sys.path:
    sys.path.insert(0, str(SRC))

from eeg.io import load_bci2k
from eeg.inspect import print_summary
from eeg.viz import plot_channels, plot_channels_psd, plot_montage

DATA_FILE = PROJECT_ROOT / "data/eeg/raw/MET000bGridFixedS001R02.dat"
```

```python
raw = load_bci2k(DATA_FILE)
print_summary(raw)
```

## Channel montage

Visualize electrode positions. Toggle `show_names` to switch between channel names and indices.

```python
plot_montage(raw, show_names=True)   # set show_names=False for channel indices
```

## Time-domain traces

Edit `CHANNELS` to any channel names from the list above.

```python
CHANNELS = ["Cz", "Fz", "Pz"]  # change to any channel names
DURATION = 10.0  # seconds to show

plot_channels(raw, CHANNELS, duration=DURATION)
```

## Power spectral density

```python
FMAX = 100.0  # Hz

plot_channels_psd(raw, CHANNELS, fmax=FMAX)
```
